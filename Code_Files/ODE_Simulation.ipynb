```python
#Importing packages for later use
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

#defining the differential equations using the f(r,t) where r = (N, C, A, X) 
#Note this can be used with nth number of variables. The size of r array needed is N x 1 ; where N is the number of variables. 


#Definition of Constants and Parameters
P = np.array((0.25, 0.75, 1.25, 1.5, 1.75))
Rbar = np.array((0.25, 0.75, 1.25, 1.5, 1.75))
m = 0.5
c = m*2

#Definition length of time
a = 0.0
b = 200
T = 20000
h = (b - a)/T
NT = 400
tpoints = np.arange(a,b,h) #uses arange to create a evenly spaced array from a to b with a step of h.

for R in Rbar:
    for p in P:
        
        png_save_path = rf"C:\Users\phili\OneDrive\Documents\Research\Figures\V5_NCAX_DiffEq/Population_Plot_P={p}_R={R}_U=1_M={m}C={c}.png"
        def f(r,t):
            N = r[0] #Defines the N value 
            C = r[1] #Defines the C values 
            A = r[2] #Defines the A values 
            X = r[3] #Defines the X values 
            fn = -p*C*N/NT + m*N*X/NT #differential equation on N
            fc = +p*C*N/NT + c*C*X/NT - R*C #differential equation on C
            fa = +R*C - A #differential equation on A
            fx = A - m*N*X/NT - c*C*X/NT #differential equation on X
            return np.array([fn,fc,fa,fx],float) #returns the 4 differential equations.
        npoints = []
        cpoints = []
        apoints = []
        xpoints = []
        
        #This creates a bin for the r bin (2x1). with the intitial value of x (on r[0]) and y (on r[1]) which is both equal to 2.
        r = np.array([399,1,0,0], float)
        
        #a for loop that runs for the whole values of t points.
        for t in tpoints:
            npoints.append(r[0]/NT) #appends the r[0] values on the xpoints array bin rabbit pop
            cpoints.append(r[1]/NT) #appends the r[1] values on the ypoints array bin fox pop
            apoints.append(r[2]/NT)
            xpoints.append(r[3]/NT)
            k1 = h*f(r,t) #uses the previous or initial values to eval the k1 area.
            k2 = h*f(r+(0.5*k1), t+(0.5*h)) #uses the k1 approximation for the k2 area approximation
            k3 = h*f(r+(0.5*k2), t+(0.5*h)) #uses the k2 approximation for the k3 area approximation
            k4 = h*f(r+k3, t+h) #uses the k3 approximation for the k4 area approximation
            r += (k1+(2*k2)+(2*k3)+k4)/6 #uses the new estimated slope to update the new r values from the k1 to k4 values
            
       
        
        plt.plot(tpoints, npoints ,linestyle = '-', color = 'green', label='Normal cell Population', linewidth=2)
        
        plt.plot(tpoints, cpoints ,linestyle = '-', color = 'maroon', label='Cancer cell Population', linewidth=2)
        
        plt.plot(tpoints, apoints ,linestyle = '-', color = 'blue', label='Apoptotic cell Population', linewidth=2)
        
        plt.plot(tpoints, xpoints, linestyle = '-',color = 'grey', label='Empty cell Population', linewidth=2)

     
        # Add labels and title
        plt.xlabel('Iterations')
        plt.ylabel('Population')
        plt.title(f'Normal, Cancerous, Apoptotic, Empty cell population over time. \np ={p}, r={R}, u=1, m={m}, c={c} ')
        
        # Add a legend
        plt.legend()
        plt.savefig(png_save_path)
       
        # Show the plot
        plt.show()
        
        df = pd.DataFrame({
                'NormalCell': npoints,
                'CancerCell': cpoints,
                'Apoptosis': apoints,
                'Empty': xpoints
            })
            
            # Save the DataFrame to CSV
        csv_save_path = rf"C:\Users\phili\OneDrive\Documents\Research\Raw Datas\V5_NCAX_DiffEq/P={p}_R={R}_U=1_M={m}C={c}.csv"
        df.to_csv(csv_save_path, index=False)
```
